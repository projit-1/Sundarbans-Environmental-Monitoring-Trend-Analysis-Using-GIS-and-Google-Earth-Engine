```python
import pandas as pd
from pathlib import Path
```

```python
file_path = Path("Sundarbans_Monthly_Temperature_2000_2025.xlsx")
```

```python
if not file_path.exists():
    raise FileNotFoundError(
        f"File not found: {file_path.resolve()}"
    )
```

```python
df = pd.read_excel(file_path)
```

```python
print("Total rows:", df.shape[0])
```

```python
print("Total columns:", df.shape[1])
```

```python
print(df.columns.tolist())
```

```python
print(df.head())
```

```python
print(df.dtypes)
```

```python
print(df.describe)
```

```python
# Missing values report

missing_report = pd.DataFrame({
    "Column_Name": df.columns,
    "Data_Type": [str(df[col].dtype) for col in df.columns],
    "Missing_Count": [df[col].isna().sum() for col in df.columns],
    "Missing_Percentage": [
        round(df[col].isna().mean() * 100, 2)
        for col in df.columns
    ],
    "Unique_Values": [df[col].nunique(dropna=True) for col in df.columns]
})
```

```python
#  Duplicate rows
duplicate_count = int(df.duplicated().sum())

duplicate_report = pd.DataFrame({
    "Metric": ["Total Rows", "Exact Duplicate Rows"],
    "Value": [len(df), duplicate_count]
})
```

```python
# 6. Save quality report
output_folder = Path("reports")
output_folder.mkdir(parents=True, exist_ok=True)

output_file = output_folder / "temperature_quality_report.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    missing_report.to_excel(
        writer, sheet_name="Missing_Values", index=False
    )
    duplicate_report.to_excel(
        writer, sheet_name="Duplicates", index=False
    )
    df.head(20).to_excel(
        writer, sheet_name="First_20_Rows", index=False
    )
```

```python
print("Exact duplicate rows:", duplicate_count)
print("Report saved to:", output_file.resolve())
```

```python
file_path = Path("Sundarbans_Monthly_Rainfall_2000_2025.xlsx")
```

```python
if not file_path.exists():
    raise FileNotFoundError(
        f"File not found: {file_path.resolve()}"
    )
```

```python
df = pd.read_excel(file_path)
```

```python
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python
print(df.columns.tolist())
```

```python
print(df.head())
```

```python
print(df.dtypes)
```

```python
#  Missing values report
missing_report = pd.DataFrame({
    "Column_Name": df.columns,
    "Data_Type": [
        str(df[col].dtype) for col in df.columns
    ],
    "Missing_Count": [
        df[col].isna().sum() for col in df.columns
    ],
    "Missing_Percentage": [
        round(df[col].isna().mean() * 100, 2)
        for col in df.columns
    ],
    "Unique_Values": [
        df[col].nunique(dropna=True) for col in df.columns
    ]
})
```

```python
#  Exact duplicate rows
duplicate_count = int(df.duplicated().sum())

duplicate_report = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Exact Duplicate Rows"
    ],
    "Value": [
        len(df),
        duplicate_count
    ]
})
```

```python
# 6. Save the report without modifying raw data
output_folder = Path("reports")
output_folder.mkdir(parents=True, exist_ok=True)

output_file = output_folder / "rainfall_quality_report.xlsx"

with pd.ExcelWriter(
    output_file, engine="openpyxl"
) as writer:

    missing_report.to_excel(
        writer,
        sheet_name="Missing_Values",
        index=False
    )

    duplicate_report.to_excel(
        writer,
        sheet_name="Duplicates",
        index=False
    )

    df.head(20).to_excel(
        writer,
        sheet_name="First_20_Rows",
        index=False
    )
```

```python
print("Exact duplicate rows:", duplicate_count)
```

```python
print("Report saved to:", output_file.resolve())
```
